```python
import pandas as pd
import json
import yaml
from pathlib import Path
```

```python
MAIN_FOLDER = Path('..', 'outputs_000.002')
# MAIN_FOLDER = Path('..', 'outputs')

for f in MAIN_FOLDER.iterdir():
    if f.is_dir() :
        print(f.name)
```

```python
folders = [f for f in MAIN_FOLDER.iterdir() if f.is_dir() and (Path(f, 'eval_results.json')).exists() ]

params = [ yaml.safe_load( Path(f, 'parameters.yaml').read_text() )
          for f in folders ]

results = list()
for f in folders:
    content = (Path(f, 'eval_results.json')).read_text()
    p = json.loads(content)
    p['_hash_id'] = f.name
    results.append(p)
```

```python
df = pd.merge(
    pd.DataFrame(params),
    pd.DataFrame(results),
    on='_hash_id'
)
```

## Tempo de treinamento

```python
folders = [f for f in MAIN_FOLDER.iterdir() if f.is_dir() and (Path(f, 'train_results.json')).exists() ]

results = list()
for f in folders:
    content = (Path(f, 'train_results.json')).read_text()
    p = json.loads(content)
    p['_hash_id'] = f.name
    results.append(p)


df_train = pd.merge(
                pd.json_normalize(results,record_prefix='metrics'),
                df[['_hash_id', 'dataset', '_timestamp']],
                on='_hash_id'
            )

df_train.columns = df_train.columns.str.replace('metrics.', '', regex=False)
df_train['train_runtime_hours'] = df_train['train_runtime'] / 3600

md = (
    df_train[
        ['_timestamp', 'dataset', 'train_runtime_hours']
    ].sort_values('_timestamp')
    .to_markdown(index=False)
)

df_train[
    ['_timestamp', 'dataset', 'train_runtime_hours']
].sort_values('_timestamp')
```
